```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl

df = pd.read_csv("/content/earthquake_1995-2023 (1).csv")

# Count earthquakes for each country
data = df["country"].value_counts().head(10)

# Normalize values for colormap
norm = mpl.colors.Normalize(vmin=data.min(), vmax=data.max())

# Choose colormap
cmap = plt.cm.viridis

# Create colors based on bar height
colors = cmap(norm(data.values))

# Create bar plot
data.plot(kind="bar", color=colors)

# Add colorbar
sm = mpl.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])

plt.colorbar(sm, ax=plt.gca(), label="Number of Earthquakes")

plt.xlabel("Country")
plt.ylabel("Number of Earthquakes")
plt.title("Top 10 Countries by Number of Earthquakes")
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,5))

plt.hist(df['magnitude'], bins=10, color='skyblue', edgecolor='black')

plt.title('Distribution of Earthquake Magnitudes')
plt.xlabel('Earthquake Magnitude')
plt.ylabel('Number of Earthquakes')

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Create magnitude groups
df['Magnitude_Group'] = pd.cut(
    df['magnitude'],
    bins=[0, 3, 5, 7, 10],
    labels=['0-3', '3-5', '5-7', '7-10']
)

# Get top 5 countries
top_countries = df['country'].value_counts().head(5).index

grouped_data = pd.crosstab(
    df[df['country'].isin(top_countries)]['country'],
    df[df['country'].isin(top_countries)]['Magnitude_Group']
)

# Plot grouped bar chart
grouped_data.plot(
    kind='bar',
    figsize=(10, 5)
)

plt.title('Earthquakes by Country and Magnitude')
plt.xlabel('Country')
plt.ylabel('Number of Earthquakes')
plt.xticks(rotation=45)
plt.legend(title='Magnitude')
plt.tight_layout()

plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd

# Create magnitude groups
df['Magnitude_Group'] = pd.cut(
    df['magnitude'],
    bins=[0, 3, 5, 7, 10],
    labels=['0-3', '3-5', '5-7', '7-10']
)

# Get top 5 countries
top_countries = df['country'].value_counts().head(5).index

# Create data for stacked bar chart
stacked_data = pd.crosstab(
    df[df['country'].isin(top_countries)]['country'],
    df[df['country'].isin(top_countries)]['Magnitude_Group']
)

# Create stacked bar chart
stacked_data.plot(
    kind='bar',
    stacked=True,
    figsize=(10, 5)
)

plt.title('Stacked Earthquakes by Country and Magnitude')
plt.xlabel('Country')
plt.ylabel('Number of Earthquakes')
plt.xticks(rotation=45)
plt.legend(title='Magnitude')
plt.tight_layout()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = df['magnitude'].dropna()

# Number of bins
no_of_bins = 10

# Find minimum and maximum
minimum = data.min()
maximum = data.max()

# Calculate bin size
bin_size = (maximum - minimum) / no_of_bins

# Calculate frequencies
frequencies, bin_edges = np.histogram(data, bins=no_of_bins)

print("Minimum value:", minimum)
print("Maximum value:", maximum)
print("Number of bins:", no_of_bins)
print("Bin size:", bin_size)

print("\nFrequencies:")
for i in range(no_of_bins):
    print(
        f"{bin_edges[i]:.2f} - {bin_edges[i+1]:.2f} : {frequencies[i]}"
    )

# Histogram
plt.figure(figsize=(10,5))
plt.hist(data, bins=no_of_bins, color='orange', edgecolor='black')

plt.title('Distribution of Earthquake Magnitudes')
plt.xlabel('Earthquake Magnitude')
plt.ylabel('Frequency')

plt.show()
```

```python
import matplotlib.pyplot as plt

country_count = df['country'].value_counts().head(5)

plt.figure(figsize=(8,6))

plt.pie(
    country_count,
    autopct='%1.1f%%',
    startangle=90,
    colors=['red', 'blue', 'green', 'orange', 'purple']
)

plt.title('Top 5 Countries with Earthquakes')

# Show country names and their colors on the side
plt.legend(
    country_count.index,
    title='Countries',
    loc='center left',
    bbox_to_anchor=(1, 0.5)
)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
plt.scatter(df['magnitude'], df['depth'], alpha=0.7, color='purple')
plt.title('Earthquake Magnitude vs. Depth')
plt.xlabel('Magnitude')
plt.ylabel('Depth (km)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
df['date_time'] = pd.to_datetime(df['date_time'], format='%d-%m-%Y %H:%M', errors='coerce')
df_sorted = df.sort_values(by='date_time')

plt.figure(figsize=(12, 6))
plt.plot(df_sorted['date_time'], df_sorted['magnitude'], color='blue', alpha=0.8)
plt.title('Earthquake Magnitude Over Time')
plt.xlabel('Date and Time')
plt.ylabel('Magnitude')
plt.grid(True, linestyle='--', alpha=0.6)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
